```python
# Import libraries

# pandas = used for data manipulation and analysis
import pandas as pd

# matplotlib = used for creating visualizations
import matplotlib.pyplot as plt

# Load the dataset
file_path = r"C:\Users\Mary\visualization\02_activities\assignments\Assignment 3\TTC LRT Delays.csv"

# Read the CSV file into a pandas DataFrame
df = pd.read_csv(file_path)

# Calculate the average delay (in minutes) grouped by day of the week.
# groupby("Day") groups all rows by the "Day" column
# ["Min Delay"] selects the delay column
# mean() calculates the average delay per day
avg_delay_by_day = df.groupby("Day")["Min Delay"].mean()

# Change the days to follow calendar order instead of alphabetical order to improves readability and interpretability.
ordered_days = [
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday", "Sunday"
]

avg_delay_by_day = avg_delay_by_day.reindex(ordered_days)

# Create a new figure object
plt.figure(figsize=(8, 5))  # Width and height in inches

# Create bar chart
avg_delay_by_day.plot(kind="bar")

# Add title
plt.title("Average LRT Delay (Minutes) by Day of Week")

# Label x-axis
plt.xlabel("Day of Week")

# Label y-axis with units
plt.ylabel("Average Delay (Minutes)")

# Rotate x-axis labels for better readability
plt.xticks(rotation=45)

# Adjust layout to prevent label cutoff
plt.tight_layout()

# Save the figure as a high-resolution PNG file (300 dpi for quality)
output_path = r"C:\Users\Mary\visualization\02_activities\assignments\Assignment 3\Visualization1_AvgDelayByDay.png"
plt.savefig(output_path, dpi=300)

# Display 
plt.show()

# Reproducibility Details
# 1.The dataset is loaded directly from file
# 2.All transformations are programed
# 3.No manual edits needed
# 4.The same output will be produced each time the code is run

```
